# Inspect, aggregate, review
Use a kernel with plugin-value-lab[science] installed. Put counts.h5ad and settings.json in the working directory. The bundled fixture is synthetic; this is not biological validation.

```python
from pathlib import Path
from value_lab.sdk import inspect_counts, aggregate_counts, verify_run
from value_lab.core import load_json
source = Path('counts.h5ad')
settings = load_json('settings.json')
inspect_counts(source)
```

```python
# Headers/configuration only; counts and donors are not validated yet.
aggregate_counts(source, output='notebook-result', dry_run=True, **settings)
```

```python
# Repeating this cell verifies before reuse.
result = aggregate_counts(source, output='notebook-result', reuse=True, **settings)
print(result.report.read_text(encoding='utf-8'))
result
```

```python
verify_run(result.directory, expected_id=result.commitment)
import anndata as ad
counts = ad.read_h5ad(result.counts, backed='r')
print(counts.shape)
counts.file.close()
```
